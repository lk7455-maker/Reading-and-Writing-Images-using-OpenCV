### Reading and Writing Images using OpenCV
This guide demonstrates how to read, display, and write images using OpenCV in a Google Colab environment.

```python
import cv2
import urllib.request
from google.colab.patches import cv2_imshow

# 1. Download a sample image using an opener with a User-Agent header to avoid 403 Forbidden
url = 'https://upload.wikimedia.org/wikipedia/commons/4/47/PNG_transparency_demonstration_1.png'
req = urllib.request.Request(url, headers={'User-Agent': 'Mozilla/5.0'})

with urllib.request.urlopen(req) as response, open('sample_input.png', 'wb') as out_file:
    out_file.write(response.read())

# 2. Read the image
# cv2.imread loads the image in BGR format by default
img = cv2.imread('sample_input.png')

# 3. Display the image
# In Colab, we use cv2_imshow instead of cv2.imshow
print("Displaying the original image:")
cv2_imshow(img)

# 4. Perform a simple operation (e.g., convert to grayscale)
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
print("Displaying the grayscale image:")
cv2_imshow(gray_img)

# 5. Write the image back to disk
cv2.imwrite('sample_output.png', gray_img)
print("Saved the grayscale image as 'sample_output.png'")
```
